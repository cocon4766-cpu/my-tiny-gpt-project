# Q6. Train a sentiment classifier

```python
import re
import torch
import torch.nn as nn
import torch.optim as optim

torch.manual_seed(42)

# =====================================================
# DATA (question 4)
# =====================================================
reviews = [
    ("I loved this movie, it was fantastic", 1),
    ("What a great film with a wonderful story", 1),
    ("Absolutely amazing acting and beautiful scenes", 1),
    ("This was a brilliant and enjoyable experience", 1),
    ("I really liked the characters and the music", 1),
    ("Such a fun and exciting movie", 1),
    ("The plot was clever and the ending was perfect", 1),
    ("Highly recommended, one of the best films ever", 1),
    ("A heartwarming story that made me smile", 1),
    ("Great performance, I enjoyed every minute", 1),
    ("I hated this movie, it was terrible", 0),
    ("What a boring film with a awful story", 0),
    ("Absolutely horrible acting and ugly scenes", 0),
    ("This was a dull and disappointing experience", 0),
    ("I really disliked the characters and the music", 0),
    ("Such a slow and annoying movie", 0),
    ("The plot was silly and the ending was bad", 0),
    ("Not recommended, one of the worst films ever", 0),
    ("A depressing story that made me angry", 0),
    ("Poor performance, I wasted my time", 0),
]

texts = [t for t, l in reviews]
labels = [l for t, l in reviews]

def tokenize(text):
    text = re.sub(r"[^a-z\s]", "", text.lower())
    return text.split()

tokenized = [tokenize(t) for t in texts]

vocab = {"<pad>": 0, "<unk>": 1}
for tokens in tokenized:
    for word in tokens:
        if word not in vocab:
            vocab[word] = len(vocab)

max_len = max(len(t) for t in tokenized)

def encode(text):
    """Turn a review string into a padded list of word numbers."""
    ids = [vocab.get(w, vocab["<unk>"]) for w in tokenize(text)][:max_len]
    return ids + [vocab["<pad>"]] * (max_len - len(ids))

X = torch.tensor([encode(t) for t in texts], dtype=torch.long)   # (20, max_len)
y = torch.tensor(labels, dtype=torch.float32)                    # floats for BCE

print("Vocabulary size:", len(vocab))
print("X shape:", X.shape, "| y shape:", y.shape)

# =====================================================
# MODEL (question 5, but with 1 output for BCE)
# =====================================================
embedding_dim = 8

model = nn.Sequential(
    nn.Embedding(len(vocab), embedding_dim, padding_idx=0),
    nn.Flatten(),
    nn.Linear(max_len * embedding_dim, 1),    # 1 score (a "logit")
)

# =====================================================
# TRAINING: BCE with logits loss + Adam
# =====================================================
criterion = nn.BCEWithLogitsLoss()
optimizer = optim.Adam(model.parameters(), lr=0.01)

epochs = 100
for epoch in range(epochs):
    model.train()
    optimizer.zero_grad()

    logits = model(X).squeeze(1)        # (20, 1) -> (20,)
    loss = criterion(logits, y)
    loss.backward()
    optimizer.step()

    if (epoch + 1) % 10 == 0:
        preds = (torch.sigmoid(logits) > 0.5).float()
        acc = (preds == y).float().mean().item()
        print(f"Epoch {epoch+1:3d}/{epochs} | Loss: {loss.item():.4f} | Accuracy: {acc*100:.1f}%")

# =====================================================
# TRY NEW REVIEWS
# =====================================================
def predict(text):
    model.eval()
    x = torch.tensor([encode(text)], dtype=torch.long)
    with torch.no_grad():
        prob = torch.sigmoid(model(x).squeeze()).item()
    label = "positive" if prob > 0.5 else "negative"
    print(f"'{text}' -> {label} ({prob*100:.1f}% positive)")

predict("I loved this great movie")
predict("This was a terrible and boring film")
```

# Q7. Use average embeddings

```python
import re
import torch
import torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt

# =====================================================
# DATA
# =====================================================
reviews = [
    ("I loved this movie, it was fantastic", 1),
    ("What a great film with a wonderful story", 1),
    ("Absolutely amazing acting and beautiful scenes", 1),
    ("This was a brilliant and enjoyable experience", 1),
    ("I really liked the characters and the music", 1),
    ("Such a fun and exciting movie", 1),
    ("The plot was clever and the ending was perfect", 1),
    ("Highly recommended, one of the best films ever", 1),
    ("A heartwarming story that made me smile", 1),
    ("Great performance, I enjoyed every minute", 1),
    ("I hated this movie, it was terrible", 0),
    ("What a boring film with a awful story", 0),
    ("Absolutely horrible acting and ugly scenes", 0),
    ("This was a dull and disappointing experience", 0),
    ("I really disliked the characters and the music", 0),
    ("Such a slow and annoying movie", 0),
    ("The plot was silly and the ending was bad", 0),
    ("Not recommended, one of the worst films ever", 0),
    ("A depressing story that made me angry", 0),
    ("Poor performance, I wasted my time", 0),
]

# New reviews the models never train on (to compare generalization)
test_reviews = [
    ("A great and wonderful movie", 1),
    ("I really enjoyed the story", 1),
    ("The music was beautiful and perfect", 1),
    ("Such an amazing film", 1),
    ("A terrible and boring movie", 0),
    ("I really disliked the story", 0),
    ("The ending was awful and bad", 0),
    ("Such a silly film", 0),
]

def tokenize(text):
    text = re.sub(r"[^a-z\s]", "", text.lower())
    return text.split()

# Vocabulary is built from the TRAINING reviews only
vocab = {"<pad>": 0, "<unk>": 1}
for text, _ in reviews:
    for word in tokenize(text):
        if word not in vocab:
            vocab[word] = len(vocab)

max_len = max(len(tokenize(t)) for t, _ in reviews)

def encode(text):
    ids = [vocab.get(w, vocab["<unk>"]) for w in tokenize(text)][:max_len]
    return ids + [vocab["<pad>"]] * (max_len - len(ids))

def make_tensors(data):
    X = torch.tensor([encode(t) for t, _ in data], dtype=torch.long)
    y = torch.tensor([l for _, l in data], dtype=torch.float32)
    return X, y

X_train, y_train = make_tensors(reviews)
X_test, y_test = make_tensors(test_reviews)
print("Vocabulary size:", len(vocab), "| max_len:", max_len)
print("Train:", X_train.shape, "| Test:", X_test.shape)

embedding_dim = 8

# =====================================================
# MODEL 1: Flatten (from question 6)
# =====================================================
def build_flatten_model():
    return nn.Sequential(
        nn.Embedding(len(vocab), embedding_dim, padding_idx=0),
        nn.Flatten(),
        nn.Linear(max_len * embedding_dim, 1),
    )

# =====================================================
# MODEL 2: Average embeddings
# =====================================================
class AvgEmbeddingClassifier(nn.Module):
    def __init__(self, vocab_size, embedding_dim):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, embedding_dim, padding_idx=0)
        self.fc = nn.Linear(embedding_dim, 1)

    def forward(self, x):
        emb = self.embedding(x)                       # (batch, max_len, dim)
        mask = (x != 0).unsqueeze(-1).float()         # 1 for real words, 0 for padding
        avg = (emb * mask).sum(dim=1) / mask.sum(dim=1).clamp(min=1)   # (batch, dim)
        return self.fc(avg)                           # (batch, 1)

# =====================================================
# Training and evaluation helpers
# =====================================================
def train_model(model, epochs=100, lr=0.01):
    criterion = nn.BCEWithLogitsLoss()
    optimizer = optim.Adam(model.parameters(), lr=lr)
    losses = []
    for epoch in range(epochs):
        model.train()
        optimizer.zero_grad()
        loss = criterion(model(X_train).squeeze(1), y_train)
        loss.backward()
        optimizer.step()
        losses.append(loss.item())
    return losses

def accuracy(model, X, y):
    model.eval()
    with torch.no_grad():
        preds = (torch.sigmoid(model(X).squeeze(1)) > 0.5).float()
    return (preds == y).float().mean().item()

def count_params(model):
    return sum(p.numel() for p in model.parameters())

# =====================================================
# Train both (same seed for a fair comparison)
# =====================================================
torch.manual_seed(42)
flatten_model = build_flatten_model()
flatten_losses = train_model(flatten_model)

torch.manual_seed(42)
avg_model = AvgEmbeddingClassifier(len(vocab), embedding_dim)
avg_losses = train_model(avg_model)

# Shape flow of the averaging model (first 4 reviews)
with torch.no_grad():
    x = X_train[:4]
    emb = avg_model.embedding(x)
    mask = (x != 0).unsqueeze(-1).float()
    avg = (emb * mask).sum(dim=1) / mask.sum(dim=1).clamp(min=1)
    print("\nAverage model shapes:")
    print("Input:", x.shape)
    print("After Embedding:", emb.shape)
    print("After Average:", avg.shape)
    print("After Linear:", avg_model.fc(avg).shape)

# =====================================================
# Compare architectures
# =====================================================
print("\n" + "=" * 55)
print(f"{'':28}{'Flatten':>13}{'Average':>13}")
print("-" * 55)
print(f"{'Parameters':28}{count_params(flatten_model):>13}{count_params(avg_model):>13}")
print(f"{'Final train loss':28}{flatten_losses[-1]:>13.4f}{avg_losses[-1]:>13.4f}")
print(f"{'Train accuracy':28}{accuracy(flatten_model, X_train, y_train)*100:>12.1f}%{accuracy(avg_model, X_train, y_train)*100:>12.1f}%")
print(f"{'Test accuracy (new reviews)':28}{accuracy(flatten_model, X_test, y_test)*100:>12.1f}%{accuracy(avg_model, X_test, y_test)*100:>12.1f}%")
print("=" * 55)

plt.plot(flatten_losses, label="Flatten")
plt.plot(avg_losses, label="Average")
plt.xlabel("Epoch"); plt.ylabel("Training loss")
plt.title("Training loss: Flatten vs Average embeddings")
plt.legend(); plt.show()
```

# Q9. Handle unseen words

```python
import re

# =====================================================
# Training reviews (the vocabulary is built from these ONLY)
# =====================================================
train_texts = [
    "I loved this movie, it was fantastic",
    "What a great film with a wonderful story",
    "Absolutely amazing acting and beautiful scenes",
    "This was a brilliant and enjoyable experience",
    "I really liked the characters and the music",
    "Such a fun and exciting movie",
    "The plot was clever and the ending was perfect",
    "Highly recommended, one of the best films ever",
    "A heartwarming story that made me smile",
    "Great performance, I enjoyed every minute",
    "I hated this movie, it was terrible",
    "What a boring film with a awful story",
    "Absolutely horrible acting and ugly scenes",
    "This was a dull and disappointing experience",
    "I really disliked the characters and the music",
    "Such a slow and annoying movie",
    "The plot was silly and the ending was bad",
    "Not recommended, one of the worst films ever",
    "A depressing story that made me angry",
    "Poor performance, I wasted my time",
]

# =====================================================
# Tokenizer and vocabulary
# =====================================================
PAD, UNK = "<PAD>", "<UNK>"

def tokenize(text):
    text = re.sub(r"[^a-z\s]", "", text.lower())
    return text.split()

vocab = {PAD: 0, UNK: 1}
for text in train_texts:
    for word in tokenize(text):
        if word not in vocab:
            vocab[word] = len(vocab)

max_len = max(len(tokenize(t)) for t in train_texts)
print("Vocabulary size:", len(vocab), "| max_len:", max_len)
print(f"{PAD} index: {vocab[PAD]} | {UNK} index: {vocab[UNK]}")

def word_to_id(word):
    """Known word -> its index. Unknown word -> the <UNK> index."""
    return vocab.get(word, vocab[UNK])

def encode(text):
    """Review string -> list of indices, cut or padded to max_len."""
    ids = [word_to_id(w) for w in tokenize(text)][:max_len]
    return ids + [vocab[PAD]] * (max_len - len(ids))

# =====================================================
# QUESTION 9: Test with sentences containing unseen words
# =====================================================
test_sentences = [
    "I loved this fantastic movie",                                           # control: all words known
    "The cinematography was breathtaking and the soundtrack was unforgettable",
    "This mediocre film felt overrated",
    "Zxqwv blorptastic movie",                                                # gibberish
    "LOVED THIS MOVIE!!!",                                                    # uppercase and punctuation
]

def show_mapping(sentence):
    print(f"\nSentence: {sentence}")
    print(f"Tokens:   {tokenize(sentence)}")
    for w in tokenize(sentence):
        status = "known" if w in vocab else f"UNKNOWN -> {UNK}"
        print(f"  {w:<18} -> {word_to_id(w):<3} ({status})")
    print("Encoded: ", encode(sentence))

for s in test_sentences:
    show_mapping(s)

# =====================================================
# Check: every unknown word maps to <UNK>, every known word does not
# =====================================================
total_unknown = 0
for s in test_sentences:
    for w in tokenize(s):
        if w in vocab:
            assert word_to_id(w) == vocab[w] and word_to_id(w) != vocab[UNK]
        else:
            assert word_to_id(w) == vocab[UNK]
            total_unknown += 1

print(f"\nAll checks passed: {total_unknown} unseen words were mapped to {UNK} (index {vocab[UNK]}).")

# The vocabulary must not grow when unseen words are encoded
size_before = len(vocab)
encode("some completely new words here")
assert len(vocab) == size_before
print("Vocabulary size unchanged after encoding unseen words:", len(vocab))
```

# Q10. Build a sentiment analysis application

```python

import random
import re
from collections import Counter
from pathlib import Path

import torch
from torch import nn

# ==========================================
# 1. CONFIGURATION
# ==========================================

random.seed(42)
torch.manual_seed(42)

MODEL_PATH = "sentiment_model.pt"

# Labels: 0 = Negative, 1 = Positive
dataset = [
    ("I love this movie", 1),
    ("This film is amazing", 1),
    ("The acting was wonderful", 1),
    ("I really enjoyed this story", 1),
    ("This product is excellent", 1),
    ("The food was delicious", 1),
    ("The service was friendly", 1),
    ("I am very happy with this", 1),
    ("This was a fantastic experience", 1),
    ("The quality is excellent", 1),
    ("I absolutely love it", 1),
    ("The movie was brilliant", 1),
    ("This product works perfectly", 1),
    ("The meal was tasty", 1),
    ("I would recommend this", 1),
    ("The experience was wonderful", 1),
    ("I am satisfied with my purchase", 1),
    ("The staff were helpful", 1),
    ("This is a great product", 1),
    ("Everything was perfect", 1),

    ("I hate this movie", 0),
    ("This film is terrible", 0),
    ("The acting was awful", 0),
    ("I really disliked this story", 0),
    ("This product is disappointing", 0),
    ("The food was disgusting", 0),
    ("The service was rude", 0),
    ("I am very unhappy with this", 0),
    ("This was a horrible experience", 0),
    ("The quality is poor", 0),
    ("I absolutely hate it", 0),
    ("The movie was boring", 0),
    ("This product barely works", 0),
    ("The meal was tasteless", 0),
    ("I would not recommend this", 0),
    ("The experience was awful", 0),
    ("I regret my purchase", 0),
    ("The staff were unhelpful", 0),
    ("This is a terrible product", 0),
    ("Everything was disappointing", 0),
]


# ==========================================
# 2. PROCESS TEXT
# ==========================================

def process_text(text):
    """Lowercase text and split it into words."""
    return re.findall(r"[a-z']+", text.lower())


# ==========================================
# 3. SPLIT DATA INTO TRAINING AND TEST SETS
# ==========================================

def split_dataset(data, test_ratio=0.25):
    positives = [item for item in data if item[1] == 1]
    negatives = [item for item in data if item[1] == 0]

    random.shuffle(positives)
    random.shuffle(negatives)

    def split_class(items):
        number_of_test_examples = max(
            1, int(len(items) * test_ratio)
        )
        return (
            items[number_of_test_examples:],
            items[:number_of_test_examples],
        )

    positive_train, positive_test = split_class(positives)
    negative_train, negative_test = split_class(negatives)

    train_data = positive_train + negative_train
    test_data = positive_test + negative_test

    random.shuffle(train_data)
    random.shuffle(test_data)

    return train_data, test_data


# ==========================================
# 4. BUILD VOCABULARY
# ==========================================

def build_vocabulary(train_data):
    # Reserve IDs for padding and unknown words.
    vocabulary = {
        "<PAD>": 0,
        "<UNK>": 1,
    }

    word_counts = Counter()

    # Build vocabulary from training text only.
    for review, label in train_data:
        words = process_text(review)
        word_counts.update(words)

    for word, count in word_counts.most_common():
        if word not in vocabulary:
            vocabulary[word] = len(vocabulary)

    return vocabulary


# ==========================================
# 5. CONVERT TEXT INTO NUMERICAL TENSORS
# ==========================================

def text_to_tensor(text, vocabulary):
    words = process_text(text)

    # Convert every word into its integer ID.
    word_ids = [
        vocabulary.get(word, vocabulary["<UNK>"])
        for word in words
    ]

    # Bag-of-words: count each word's occurrences.
    vector = torch.zeros(
        len(vocabulary),
        dtype=torch.float32
    )

    for word_id in word_ids:
        vector[word_id] += 1.0

    return vector


def prepare_batch(data, vocabulary):
    features = torch.stack([
        text_to_tensor(review, vocabulary)
        for review, label in data
    ])

    labels = torch.tensor(
        [label for review, label in data],
        dtype=torch.long
    )

    return features, labels


# ==========================================
# 6. BUILD THE NEURAL NETWORK
# ==========================================

class SentimentNetwork(nn.Module):

    def __init__(self, vocabulary_size):
        super().__init__()

        self.network = nn.Sequential(
            nn.Linear(vocabulary_size, 32),
            nn.ReLU(),
            nn.Linear(32, 2)
        )

    def forward(self, x):
        return self.network(x)


# ==========================================
# 7. TRAIN THE MODEL
# ==========================================

def train_model(epochs=100):
    train_data, test_data = split_dataset(dataset)

    vocabulary = build_vocabulary(train_data)

    train_features, train_labels = prepare_batch(
        train_data, vocabulary
    )

    model = SentimentNetwork(len(vocabulary))

    loss_function = nn.CrossEntropyLoss()

    optimizer = torch.optim.Adam(
        model.parameters(),
        lr=0.03
    )

    print("Training the sentiment model...")

    for epoch in range(epochs):
        model.train()

        # Forward pass: make predictions.
        outputs = model(train_features)

        # Calculate prediction error.
        loss = loss_function(outputs, train_labels)

        # Clear previous gradients.
        optimizer.zero_grad()

        # Backpropagation.
        loss.backward()

        # Update neural network weights.
        optimizer.step()

        if (epoch + 1) % 20 == 0:
            print(
                f"Epoch {epoch + 1}/{epochs}, "
                f"Loss: {loss.item():.4f}"
            )

    # ======================================
    # 8. EVALUATE ON HELD-OUT TEST DATA
    # ======================================

    test_features, test_labels = prepare_batch(
        test_data, vocabulary
    )

    model.eval()

    with torch.no_grad():
        test_outputs = model(test_features)
        predictions = test_outputs.argmax(dim=1)

        correct = (predictions == test_labels).sum().item()
        total = len(test_labels)
        accuracy = correct / total

    print("\nEvaluation results")
    print("------------------")
    print(f"Correct predictions: {correct}/{total}")
    print(f"Test accuracy: {accuracy:.2%}")

    # ======================================
    # 9. SAVE MODEL AND VOCABULARY
    # ======================================

    torch.save(
        {
            "model_state_dict": model.state_dict(),
            "vocabulary": vocabulary,
            "vocabulary_size": len(vocabulary),
        },
        MODEL_PATH
    )

    print(f"\nModel saved to: {MODEL_PATH}")

    return model, vocabulary


# ==========================================
# 10. LOAD A SAVED MODEL
# ==========================================

def load_model():
    if not Path(MODEL_PATH).exists():
        raise FileNotFoundError(
            "Model file not found. Train the model first."
        )

    # Only load checkpoints you created or trust.
    checkpoint = torch.load(
        MODEL_PATH,
        map_location="cpu",
        weights_only=True
    )

    vocabulary = checkpoint["vocabulary"]

    model = SentimentNetwork(
        checkpoint["vocabulary_size"]
    )

    model.load_state_dict(
        checkpoint["model_state_dict"]
    )

    model.eval()

    return model, vocabulary


# ==========================================
# 11. PREDICT SENTIMENT
# ==========================================

def predict_sentiment(text, model, vocabulary):
    features = text_to_tensor(text, vocabulary)

    # Add a batch dimension.
    features = features.unsqueeze(0)

    model.eval()

    with torch.no_grad():
        outputs = model(features)

        probabilities = torch.softmax(outputs, dim=1)[0]

        predicted_label = int(
            probabilities.argmax().item()
        )

        confidence = probabilities[predicted_label].item()

    sentiment = (
        "Positive" if predicted_label == 1
        else "Negative"
    )

    return sentiment, confidence


# ==========================================
# 12. EXAMPLE PREDICTIONS
# ==========================================

def show_examples(model, vocabulary):
    examples = [
        "I love this amazing movie",
        "This product is terrible",
        "The food was delicious",
        "The service was rude",
        "I really enjoyed this",
        "Everything was disappointing",
    ]

    print("\nExample predictions")
    print("-------------------")

    for review in examples:
        sentiment, confidence = predict_sentiment(
            review, model, vocabulary
        )

        print(f"Review: {review}")
        print(f"Prediction: {sentiment}")
        print(f"Model score: {confidence:.2%}")
        print()


# ==========================================
# 13. MAIN APPLICATION
# ==========================================

if __name__ == "__main__":

    print("SENTIMENT ANALYSIS APPLICATION")
    print("==============================")

    if Path(MODEL_PATH).exists():
        print("Loading the saved model...")
        model, vocabulary = load_model()

    else:
        print("No saved model found.")
        model, vocabulary = train_model()

        # Reload to verify that saving/loading works.
        model, vocabulary = load_model()
        print("Saved model reloaded successfully.")

    show_examples(model, vocabulary)

    print("Enter a review to analyse.")
    print("Type 'quit' to stop.\n")

    while True:
        review = input("Your review: ").strip()

        if review.lower() == "quit":
            print("Goodbye!")
            break

        if not review:
            print("Please enter a review.")
            continue

        sentiment, confidence = predict_sentiment(
            review, model, vocabulary
        )

        print(f"Predicted sentiment: {sentiment}")
        print(f"Model score: {confidence:.2%}\n")
```

# Q7. Replace the RNN with an LSTM

```python
import torch
import torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader, random_split

torch.manual_seed(42)

# 1. Generate the same synthetic dataset
X = torch.randn(1000, 10, 4)
y = (X[:, :, 0].mean(dim=1) > 0).long()

dataset = TensorDataset(X, y)
train_dataset, test_dataset = random_split(
    dataset, [800, 200],
    generator=torch.Generator().manual_seed(42)
)

train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=32)


# 2. RNN classifier
class RNNClassifier(nn.Module):
    def __init__(self):
        super().__init__()
        self.rnn = nn.RNN(4, 32, batch_first=True)
        self.fc = nn.Linear(32, 3)

    def forward(self, x):
        output, hidden = self.rnn(x)
        return self.fc(hidden[-1])


# 3. LSTM classifier
class LSTMClassifier(nn.Module):
    def __init__(self):
        super().__init__()
        self.lstm = nn.LSTM(4, 32, batch_first=True)
        self.fc = nn.Linear(32, 3)

    def forward(self, x):
        output, (hidden, cell) = self.lstm(x)
        return self.fc(hidden[-1])


# 4. Training and evaluation function
def train_and_evaluate(model, name):
    criterion = nn.CrossEntropyLoss()
    optimizer = torch.optim.Adam(model.parameters(), lr=0.001)

    for epoch in range(10):
        model.train()

        for sequences, labels in train_loader:
            optimizer.zero_grad()
            predictions = model(sequences)
            loss = criterion(predictions, labels)
            loss.backward()
            optimizer.step()

    model.eval()
    correct = 0
    total = 0

    with torch.no_grad():
        for sequences, labels in test_loader:
            predictions = model(sequences)
            predicted_classes = predictions.argmax(dim=1)

            correct += (predicted_classes == labels).sum().item()
            total += labels.size(0)

    accuracy = 100 * correct / total
    print(f"{name} Test Accuracy: {accuracy:.2f}%")

    return accuracy


# 5. Train both models on the same split
torch.manual_seed(100)
rnn_model = RNNClassifier()
rnn_accuracy = train_and_evaluate(rnn_model, "RNN")

torch.manual_seed(100)
lstm_model = LSTMClassifier()
lstm_accuracy = train_and_evaluate(lstm_model, "LSTM")

print("\nComparison:")
print(f"RNN:  {rnn_accuracy:.2f}%")
print(f"LSTM: {lstm_accuracy:.2f}%")
print(f"Difference (LSTM - RNN): {lstm_accuracy - rnn_accuracy:.2f}%")
```

```python
import torch
import torch.nn as nn
import math

torch.manual_seed(42)

# 1. Generate sine wave data
t = torch.linspace(0, 100, 3000)
data = torch.sin(t)

# 2. Create sequences: previous 20 values -> next value
sequence_length = 20
X, y = [], []

for i in range(len(data) - sequence_length):
    X.append(data[i:i + sequence_length])
    y.append(data[i + sequence_length])

X = torch.stack(X).unsqueeze(-1)
y = torch.stack(y).unsqueeze(-1)

# 3. Split into training and testing sets
split = int(0.8 * len(X))

X_train, X_test = X[:split], X[split:]
y_train, y_test = y[:split], y[split:]


# 4. Build LSTM regression model
class LSTMRegressor(nn.Module):
    def __init__(self):
        super().__init__()
        self.lstm = nn.LSTM(
            input_size=1,
            hidden_size=32,
            batch_first=True
        )
        self.fc = nn.Linear(32, 1)

    def forward(self, x):
        output, (hidden, cell) = self.lstm(x)
        return self.fc(hidden[-1])


model = LSTMRegressor()

# 5. Loss and optimizer
criterion = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)

# 6. Train the model
epochs = 20
batch_size = 32

for epoch in range(epochs):
    model.train()
    total_loss = 0

    for i in range(0, len(X_train), batch_size):
        batch_X = X_train[i:i + batch_size]
        batch_y = y_train[i:i + batch_size]

        optimizer.zero_grad()
        predictions = model(batch_X)
        loss = criterion(predictions, batch_y)
        loss.backward()
        optimizer.step()

        total_loss += loss.item()

    print(
        f"Epoch {epoch + 1}/{epochs}, "
        f"Loss: {total_loss / math.ceil(len(X_train) / batch_size):.6f}"
    )

# 7. Evaluate on held-out test data
model.eval()

with torch.no_grad():
    test_predictions = model(X_test)
    test_loss = criterion(test_predictions, y_test)

print(f"\nTest MSE: {test_loss.item():.6f}")

# 8. Compare several predictions
print("\nActual vs Predicted:")
for i in range(10):
    print(
        f"Actual: {y_test[i].item():.4f}, "
        f"Predicted: {test_predictions[i].item():.4f}"
    )
```

# Q10. Time-Series Prediction

```python
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
from torch.utils.data import TensorDataset, DataLoader

torch.manual_seed(42)

# ==========================================
# 1. Generate time series data
# ==========================================
time = torch.linspace(0, 100, 3000)
data = torch.sin(time)

# ==========================================
# 2. Create input sequences and targets
# Use previous 20 values to predict next value
# ==========================================
sequence_length = 20
split_index = int(len(data) * 0.8)


def create_sequences(series, start, end, window):
    X, y = [], []

    for i in range(start, end):
        X.append(series[i - window:i])
        y.append(series[i])

    X = torch.stack(X).unsqueeze(-1)
    y = torch.stack(y).unsqueeze(-1)

    return X, y


# ==========================================
# 3. Chronological train/test split
# ==========================================
X_train, y_train = create_sequences(
    data, sequence_length, split_index, sequence_length
)

X_test, y_test = create_sequences(
    data, split_index, len(data), sequence_length
)

train_loader = DataLoader(
    TensorDataset(X_train, y_train),
    batch_size=32,
    shuffle=True
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
print("Input shape:", X_train.shape)

# ==========================================
# 4. Build the LSTM regression model
# ==========================================
class LSTMRegressor(nn.Module):
    def __init__(self):
        super().__init__()

        self.lstm = nn.LSTM(
            input_size=1,
            hidden_size=32,
            batch_first=True
        )

        self.fc = nn.Linear(32, 1)

    def forward(self, x):
        output, (hidden, cell) = self.lstm(x)
        final_hidden = hidden[-1]
        prediction = self.fc(final_hidden)

        return prediction


model = LSTMRegressor()

criterion = nn.MSELoss()
optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.001
)

# ==========================================
# 5. Train the model
# ==========================================
epochs = 30

for epoch in range(epochs):
    model.train()
    total_loss = 0

    for sequences, targets in train_loader:
        optimizer.zero_grad()

        predictions = model(sequences)
        loss = criterion(predictions, targets)

        loss.backward()
        optimizer.step()

        total_loss += loss.item()

    average_loss = total_loss / len(train_loader)

    print(
        f"Epoch [{epoch + 1}/{epochs}] "
        f"Loss: {average_loss:.6f}"
    )

# ==========================================
# 6. Save the trained model
# ==========================================
torch.save(model.state_dict(), "lstm_time_series.pth")
print("\nModel saved successfully!")

# ==========================================
# 7. Reload the saved model
# ==========================================
loaded_model = LSTMRegressor()

loaded_model.load_state_dict(
    torch.load(
        "lstm_time_series.pth",
        map_location="cpu",
        weights_only=True
    )
)

loaded_model.eval()
print("Model reloaded successfully!")

# ==========================================
# 8. Evaluate on the held-out test set
# ==========================================
with torch.no_grad():
    test_predictions = loaded_model(X_test)
    test_mse = criterion(test_predictions, y_test)

print(f"\nTest MSE: {test_mse.item():.6f}")

# ==========================================
# 9. Plot actual versus predicted values
# ==========================================
actual = y_test.squeeze().numpy()
predicted = test_predictions.squeeze().numpy()
test_time = time[split_index:].numpy()

plt.figure(figsize=(12, 5))

plt.plot(test_time, actual, label="Actual Values")
plt.plot(test_time, predicted, label="Predicted Values")

plt.title("LSTM Time Series Prediction")
plt.xlabel("Time")
plt.ylabel("Value")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

# Q6. Add an attention mask

```python
import torch
import math

def scaled_dot_product_attention(Q, K, V):
    d_k = K.shape[-1]

    # Calculate attention scores
    scores = torch.matmul(Q, K.transpose(-2, -1))
    scores = scores / math.sqrt(d_k)

    # Create causal mask: prevent attending to future positions
    seq_len = Q.shape[-2]
    mask = torch.triu(
        torch.ones(seq_len, K.shape[-2], dtype=torch.bool),
        diagonal=1
    ).to(Q.device)

    # Mask future positions before softmax
    scores = scores.masked_fill(mask, float("-inf"))

    # Calculate attention weights and output
    weights = torch.softmax(scores, dim=-1)
    output = torch.matmul(weights, V)

    return output, weights


# Test with a sequence of 4 positions
Q = torch.rand(1, 4, 4)
K = torch.rand(1, 4, 4)
V = torch.rand(1, 4, 6)

output, weights = scaled_dot_product_attention(Q, K, V)

print("Attention weights:")
print(weights)

print("\nOutput shape:", output.shape)

# Verify future positions receive zero attention
print("\nFuture attention weights are zero:")
print(torch.allclose(
    weights[0].triu(diagonal=1),
    torch.zeros_like(weights[0].triu(diagonal=1))
))
```

# Q7. Build a multi-head attention

```python
import torch
import torch.nn as nn
import math

class MultiHeadAttention(nn.Module):
    def __init__(self, input_dim=32, num_heads=4):
        super().__init__()

        self.num_heads = num_heads
        self.head_dim = input_dim // num_heads

        self.query = nn.Linear(input_dim, input_dim)
        self.key = nn.Linear(input_dim, input_dim)
        self.value = nn.Linear(input_dim, input_dim)

        self.fc_out = nn.Linear(input_dim, input_dim)

    def forward(self, x):
        batch_size, seq_length, input_dim = x.shape

        # Create queries, keys, and values
        Q = self.query(x)
        K = self.key(x)
        V = self.value(x)

        # Split into 4 attention heads
        Q = Q.reshape(batch_size, seq_length, self.num_heads, self.head_dim)
        K = K.reshape(batch_size, seq_length, self.num_heads, self.head_dim)
        V = V.reshape(batch_size, seq_length, self.num_heads, self.head_dim)

        # Move heads dimension before sequence length
        Q = Q.transpose(1, 2)
        K = K.transpose(1, 2)
        V = V.transpose(1, 2)

        # Scaled dot-product attention
        scores = torch.matmul(Q, K.transpose(-2, -1))
        scores = scores / math.sqrt(self.head_dim)

        weights = torch.softmax(scores, dim=-1)
        output = torch.matmul(weights, V)

        # Combine the heads
        output = output.transpose(1, 2).contiguous()
        output = output.reshape(batch_size, seq_length, input_dim)

        return self.fc_out(output)


# Test the module
model = MultiHeadAttention(input_dim=32, num_heads=4)

x = torch.rand(2, 10, 32)
output = model(x)

print("Input shape:", x.shape)
print("Output shape:", output.shape)
```

```python
import torch
import torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader, random_split

torch.manual_seed(42)

# 1. Generate synthetic sequential classification data
num_samples = 1000
seq_length = 10
input_dim = 4
num_classes = 3

X = torch.randn(num_samples, seq_length, input_dim)

# Label each sequence based on its mean value
scores = X[:, :, 0].mean(dim=1)
y = torch.zeros(num_samples, dtype=torch.long)
y[scores > 0.3] = 1
y[scores < -0.3] = 2

dataset = TensorDataset(X, y)

train_dataset, test_dataset = random_split(
    dataset,
    [800, 200],
    generator=torch.Generator().manual_seed(42)
)

train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=32)


# 2. Build attention-based sequence classifier
class AttentionClassifier(nn.Module):
    def __init__(self):
        super().__init__()

        # Input projection: 4 features -> 32 dimensions
        self.projection = nn.Linear(4, 32)

        # Multi-head self-attention
        self.attention = nn.MultiheadAttention(
            embed_dim=32,
            num_heads=4,
            batch_first=True
        )

        # Feed-forward layer
        self.feed_forward = nn.Sequential(
            nn.Linear(32, 64),
            nn.ReLU(),
            nn.Linear(64, 32)
        )

        # Final classification layer
        self.classifier = nn.Linear(32, 3)

    def forward(self, x):
        x = self.projection(x)

        attention_output, _ = self.attention(x, x, x)
        x = x + attention_output

        x = x + self.feed_forward(x)

        # Combine information across all time steps
        x = x.mean(dim=1)

        return self.classifier(x)


model = AttentionClassifier()

# 3. Set up training
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)

# 4. Train the model
epochs = 15

for epoch in range(epochs):
    model.train()
    total_loss = 0

    for sequences, labels in train_loader:
        optimizer.zero_grad()

        predictions = model(sequences)
        loss = criterion(predictions, labels)

        loss.backward()
        optimizer.step()

        total_loss += loss.item()

    print(
        f"Epoch {epoch + 1}/{epochs}, "
        f"Loss: {total_loss / len(train_loader):.4f}"
    )


# 5. Evaluate on test data
model.eval()
correct = 0
total = 0

with torch.no_grad():
    for sequences, labels in test_loader:
        predictions = model(sequences)
        predicted_classes = predictions.argmax(dim=1)

        correct += (predicted_classes == labels).sum().item()
        total += labels.size(0)

accuracy = 100 * correct / total

print(f"\nTest Accuracy: {accuracy:.2f}%")
```

# Q10. Build a tiny transformer block

```python
import torch
import torch.nn as nn

class TransformerBlock(nn.Module):
    def __init__(self, embed_dim=32, num_heads=4):
        super().__init__()

        # 1. Multi-head self-attention
        self.attention = nn.MultiheadAttention(
            embed_dim=embed_dim,
            num_heads=num_heads,
            batch_first=True
        )

        # 2. First layer normalization
        self.norm1 = nn.LayerNorm(embed_dim)

        # 3. Feed-forward network
        self.feed_forward = nn.Sequential(
            nn.Linear(embed_dim, 128),
            nn.ReLU(),
            nn.Linear(128, embed_dim)
        )

        # 4. Second layer normalization
        self.norm2 = nn.LayerNorm(embed_dim)

    def forward(self, x):
        # Self-attention
        attention_output, _ = self.attention(x, x, x)

        # First residual connection + normalization
        x = self.norm1(x + attention_output)

        # Feed-forward network
        ff_output = self.feed_forward(x)

        # Second residual connection + normalization
        x = self.norm2(x + ff_output)

        return x


# Test the Transformer block
model = TransformerBlock()

x = torch.rand(16, 10, 32)

output = model(x)

print("Input shape:", x.shape)
print("Output shape:", output.shape)
```

```python
# Question 6: Transformer Block

import torch
import torch.nn as nn

class TransformerBlock(nn.Module):
    def __init__(self, embedding_dim=64, num_heads=4):
        super().__init__()

        # 1. Multi-head self-attention
        self.attention = nn.MultiheadAttention(
            embed_dim=embedding_dim,
            num_heads=num_heads,
            batch_first=True
        )

        # 3. First layer normalization
        self.norm1 = nn.LayerNorm(embedding_dim)

        # 4. Feedforward network
        self.feedforward = nn.Sequential(
            nn.Linear(embedding_dim, 256),
            nn.ReLU(),
            nn.Linear(256, embedding_dim)
        )

        # Layer normalization after the feedforward network
        self.norm2 = nn.LayerNorm(embedding_dim)

    def forward(self, x, causal_mask):
        # 1. Multi-head self-attention
        attention_output, _ = self.attention(
            x, x, x,
            attn_mask=causal_mask,
            need_weights=False
        )

        # 2. First residual connection
        x = self.norm1(x + attention_output)

        # 4. Feedforward network
        ff_output = self.feedforward(x)

        # 5. Second residual connection
        x = self.norm2(x + ff_output)

        return x


# Test the transformer block
batch_size = 4
sequence_length = 32
embedding_dim = 64

x = output  # Token + positional embeddings

causal_mask = torch.triu(
    torch.ones(sequence_length, sequence_length, dtype=torch.bool),
    diagonal=1
)

block = TransformerBlock(embedding_dim=64, num_heads=4)

result = block(x, causal_mask)

print("Input shape:", x.shape)
print("Output shape:", result.shape)
```

# Q7. Build Mini GPT combine

```python
# Question 7: Build a Mini GPT

import torch
import torch.nn as nn

class MiniGPT(nn.Module):
    def __init__(self, vocab_size, embedding_dim=64, num_heads=4,
                 max_sequence_length=32):
        super().__init__()

        # 1. Token embeddings
        self.token_embedding = nn.Embedding(vocab_size, embedding_dim)

        # 2. Positional embeddings
        self.position_embedding = nn.Embedding(
            max_sequence_length, embedding_dim
        )

        # 3. Two Transformer blocks
        self.blocks = nn.ModuleList([
            TransformerBlock(embedding_dim, num_heads),
            TransformerBlock(embedding_dim, num_heads)
        ])

        # 4. Final layer normalization
        self.final_norm = nn.LayerNorm(embedding_dim)

        # 5. Linear output layer: predict next token
        self.output_layer = nn.Linear(embedding_dim, vocab_size)

        self.max_sequence_length = max_sequence_length

    def forward(self, x):
        batch_size, sequence_length = x.shape

        # Token + positional embeddings
        token_emb = self.token_embedding(x)
        positions = torch.arange(sequence_length, device=x.device)
        pos_emb = self.position_embedding(positions)

        x = token_emb + pos_emb.unsqueeze(0)

        # Causal mask: block access to future tokens
        causal_mask = torch.triu(
            torch.ones(
                sequence_length, sequence_length,
                device=x.device, dtype=torch.bool
            ),
            diagonal=1
        )

        # Pass through two Transformer blocks
        for block in self.blocks:
            x = block(x, causal_mask)

        # Final normalization and next-token predictions
        x = self.final_norm(x)
        logits = self.output_layer(x)

        return logits


# Test the Mini GPT
V = len(vocab)

model = MiniGPT(vocab_size=V)

# Input shape: (batch_size, sequence_length)
sample_input = torch.randint(0, V, (4, 32))

# Output shape: (batch_size, sequence_length, vocab_size)
logits = model(sample_input)

print("Input shape:", sample_input.shape)
print("Output shape:", logits.shape)
print("Vocabulary size:", V)
```

```python

import torch

def generate_text(model, prompt, temperature=1.0, max_new_tokens=100):
    model.eval()

    input_ids = torch.tensor(
        [encode(prompt)],
        dtype=torch.long,
        device=device
    )

    with torch.no_grad():
        for _ in range(max_new_tokens):
            # Keep only the most recent context
            context = input_ids[:, -model.max_sequence_length:]

            # Get predictions for the next character
            logits = model(context)
            next_token_logits = logits[:, -1, :]

            # Apply temperature
            next_token_logits = next_token_logits / temperature

            # Convert scores into probabilities
            probabilities = torch.softmax(next_token_logits, dim=-1)

            # Sample the next character
            next_token = torch.multinomial(probabilities, num_samples=1)

            # Add it to the sequence
            input_ids = torch.cat([input_ids, next_token], dim=1)

    return decode(input_ids[0].tolist())
```

# Q5. Top-k sampling

```python

import torch

def generate_text(
    model,
    prompt,
    temperature=1.0,
    top_k=None,
    max_new_tokens=100
):
    model.eval()

    input_ids = torch.tensor(
        [encode(prompt)],
        dtype=torch.long,
        device=device
    )

    with torch.no_grad():
        for _ in range(max_new_tokens):
            context = input_ids[:, -model.max_sequence_length:]

            logits = model(context)
            next_token_logits = logits[:, -1, :]

            # Apply temperature
            next_token_logits = next_token_logits / temperature

            # Apply Top-K filtering if requested
            if top_k is not None:
                k = min(top_k, next_token_logits.size(-1))

                top_values, _ = torch.topk(next_token_logits, k=k)

                threshold = top_values[:, -1].unsqueeze(-1)

                next_token_logits[
                    next_token_logits < threshold
                ] = float("-inf")

            # Convert logits to probabilities
            probabilities = torch.softmax(next_token_logits, dim=-1)

            # Sample the next character
            next_token = torch.multinomial(probabilities, num_samples=1)

            input_ids = torch.cat([input_ids, next_token], dim=1)

    return decode(input_ids[0].tolist())
```

# Q7. Context length experiment

```python

import time
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader

# Use the same train/validation split for both models
split_index = int(0.9 * len(token_ids))
train_data = token_ids[:split_index]
val_data = token_ids[split_index:]

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Keep settings the same for both experiments
EMBEDDING_DIM = 64
NUM_HEADS = 4
BATCH_SIZE = 16
EPOCHS = 100
LEARNING_RATE = 0.001
V = len(vocab)

def create_sequences(data, context_length):
    inputs, targets = [], []

    for i in range(len(data) - context_length):
        inputs.append(data[i:i + context_length])
        targets.append(data[i + 1:i + context_length + 1])

    return torch.stack(inputs), torch.stack(targets)


def evaluate_validation(model, context_length):
    model.eval()
    total_loss = 0.0
    count = 0

    # Predict each validation character using only its preceding context
    with torch.no_grad():
        for i in range(len(val_data)):
            previous = torch.cat([
                train_data[-context_length:],
                val_data[:i]
            ])

            context = previous[-context_length:].unsqueeze(0).to(device)
            target = val_data[i].view(1).to(device)

            logits = model(context)
            loss = nn.functional.cross_entropy(
                logits[:, -1, :], target
            )

            total_loss += loss.item()
            count += 1

    return total_loss / count


def run_experiment(context_length):
    # Same seed makes the initializations more comparable
    torch.manual_seed(42)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(42)
        torch.cuda.reset_peak_memory_stats()

    X, Y = create_sequences(train_data, context_length)

    loader_generator = torch.Generator().manual_seed(123)
    dataset = TensorDataset(X, Y)
    loader = DataLoader(
        dataset,
        batch_size=BATCH_SIZE,
        shuffle=True,
        generator=loader_generator
    )

    model = MiniGPT(
        vocab_size=V,
        embedding_dim=EMBEDDING_DIM,
        num_heads=NUM_HEADS,
        max_sequence_length=context_length
    ).to(device)

    criterion = nn.CrossEntropyLoss()
    optimizer = optim.Adam(model.parameters(), lr=LEARNING_RATE)

    start_time = time.perf_counter()

    for epoch in range(EPOCHS):
        model.train()
        total_loss = 0.0

        for xb, yb in loader:
            xb, yb = xb.to(device), yb.to(device)

            optimizer.zero_grad()
            logits = model(xb)

            loss = criterion(
                logits.reshape(-1, V),
                yb.reshape(-1)
            )

            loss.backward()
            optimizer.step()
            total_loss += loss.item()

    if device.type == "cuda":
        torch.cuda.synchronize()

    training_seconds = time.perf_counter() - start_time
    train_loss = total_loss / len(loader)
    val_loss = evaluate_validation(model, context_length)

    if device.type == "cuda":
        peak_memory_mb = (
            torch.cuda.max_memory_allocated() / (1024 ** 2)
        )
    else:
        peak_memory_mb = None

    return {
        "context": context_length,
        "train_loss": train_loss,
        "val_loss": val_loss,
        "seconds": training_seconds,
        "gpu_memory_mb": peak_memory_mb
    }


results = []

for length in [32, 64]:
    print(f"\nTraining model with context length {length}...")
    result = run_experiment(length)
    results.append(result)

    print(f"Training loss: {result['train_loss']:.4f}")
    print(f"Validation loss: {result['val_loss']:.4f}")
    print(f"Training time: {result['seconds']:.2f} seconds")

    if result["gpu_memory_mb"] is not None:
        print(f"Peak GPU memory: {result['gpu_memory_mb']:.2f} MB")
    else:
        print("Peak GPU memory: Not applicable (running on CPU)")

print("\n=== FINAL COMPARISON ===")
print(f"{'Context':<12}{'Train Loss':<14}{'Val Loss':<14}"
      f"{'Time (s)':<14}{'GPU Memory MB':<14}")

for r in results:
    memory = (
        f"{r['gpu_memory_mb']:.2f}"
        if r["gpu_memory_mb"] is not None else "N/A"
    )
    print(
        f"{r['context']:<12}{r['train_loss']:<14.4f}"
        f"{r['val_loss']:<14.4f}{r['seconds']:<14.2f}{memory:<14}"
    )
```

# Q8. Model-size experiment

```python

import time
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader

# ---------------------------------
# 1. Model architecture
# ---------------------------------

class TransformerBlock(nn.Module):
    def __init__(self, embedding_dim, num_heads=4):
        super().__init__()

        self.attention = nn.MultiheadAttention(
            embedding_dim, num_heads, batch_first=True
        )
        self.norm1 = nn.LayerNorm(embedding_dim)

        self.feedforward = nn.Sequential(
            nn.Linear(embedding_dim, 4 * embedding_dim),
            nn.ReLU(),
            nn.Linear(4 * embedding_dim, embedding_dim)
        )
        self.norm2 = nn.LayerNorm(embedding_dim)

    def forward(self, x, causal_mask):
        attn_output, _ = self.attention(
            x, x, x, attn_mask=causal_mask, need_weights=False
        )
        x = self.norm1(x + attn_output)
        x = self.norm2(x + self.feedforward(x))
        return x


class MiniGPT(nn.Module):
    def __init__(
        self,
        vocab_size,
        embedding_dim=64,
        num_heads=4,
        max_sequence_length=32,
        num_blocks=2
    ):
        super().__init__()

        self.token_embedding = nn.Embedding(vocab_size, embedding_dim)
        self.position_embedding = nn.Embedding(
            max_sequence_length, embedding_dim
        )

        self.blocks = nn.ModuleList([
            TransformerBlock(embedding_dim, num_heads)
            for _ in range(num_blocks)
        ])

        self.final_norm = nn.LayerNorm(embedding_dim)
        self.output_layer = nn.Linear(embedding_dim, vocab_size)
        self.max_sequence_length = max_sequence_length

    def forward(self, x):
        _, seq_len = x.shape

        positions = torch.arange(seq_len, device=x.device)
        x = self.token_embedding(x) + self.position_embedding(
            positions
        ).unsqueeze(0)

        mask = torch.triu(
            torch.ones(
                seq_len, seq_len,
                dtype=torch.bool, device=x.device
            ),
            diagonal=1
        )

        for block in self.blocks:
            x = block(x, mask)

        return self.output_layer(self.final_norm(x))


# ---------------------------------
# 2. Consistent dataset and settings
# ---------------------------------

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Reuse the same split for both models
split_index = int(0.9 * len(token_ids))
train_data = token_ids[:split_index]
val_data = token_ids[split_index:]

CONTEXT_LENGTH = 32
BATCH_SIZE = 8
EPOCHS = 50
LEARNING_RATE = 0.001
V = len(vocab)

def make_sequences(data, context_length):
    X, Y = [], []
    for i in range(len(data) - context_length):
        X.append(data[i:i + context_length])
        Y.append(data[i + 1:i + context_length + 1])
    return torch.stack(X), torch.stack(Y)

X_train, Y_train = make_sequences(train_data, CONTEXT_LENGTH)

# Validation predicts each held-out character using prior context
def calculate_val_loss(model):
    model.eval()
    total_loss = 0.0

    with torch.no_grad():
        for i in range(len(val_data)):
            history = torch.cat([
                train_data[-CONTEXT_LENGTH:],
                val_data[:i]
            ])
            context = history[-CONTEXT_LENGTH:].unsqueeze(0).to(device)
            target = val_data[i].view(1).to(device)

            logits = model(context)
            total_loss += nn.functional.cross_entropy(
                logits[:, -1, :], target
            ).item()

    return total_loss / len(val_data)


# ---------------------------------
# 3. Training and measurement
# ---------------------------------

def run_model(name, blocks, embedding_dim):
    torch.manual_seed(42)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(42)
        torch.cuda.reset_peak_memory_stats()

    model = MiniGPT(
        vocab_size=V,
        embedding_dim=embedding_dim,
        num_heads=4,
        max_sequence_length=CONTEXT_LENGTH,
        num_blocks=blocks
    ).to(device)

    parameter_count = sum(p.numel() for p in model.parameters())

    loader = DataLoader(
        TensorDataset(X_train, Y_train),
        batch_size=BATCH_SIZE,
        shuffle=True,
        generator=torch.Generator().manual_seed(123)
    )

    optimizer = optim.Adam(model.parameters(), lr=LEARNING_RATE)
    criterion = nn.CrossEntropyLoss()

    if device.type == "cuda":
        torch.cuda.synchronize()
    start = time.perf_counter()

    for epoch in range(EPOCHS):
        model.train()
        total_loss = 0.0

        for xb, yb in loader:
            xb, yb = xb.to(device), yb.to(device)

            optimizer.zero_grad()
            logits = model(xb)
            loss = criterion(
                logits.reshape(-1, V),
                yb.reshape(-1)
            )
            loss.backward()
            optimizer.step()
            total_loss += loss.item()

        if (epoch + 1) % 10 == 0:
            print(
                f"{name} | Epoch {epoch + 1}/{EPOCHS} | "
                f"Training loss: {total_loss / len(loader):.4f}"
            )

    if device.type == "cuda":
        torch.cuda.synchronize()

    elapsed = time.perf_counter() - start
    val_loss = calculate_val_loss(model)

    peak_memory = (
        torch.cuda.max_memory_allocated() / (1024 ** 2)
        if device.type == "cuda" else None
    )

    return {
        "model": name,
        "blocks": blocks,
        "embedding_dim": embedding_dim,
        "parameters": parameter_count,
        "val_loss": val_loss,
        "seconds": elapsed,
        "gpu_memory_mb": peak_memory
    }


# ---------------------------------
# 4. Run both models
# ---------------------------------

results = []

print("Device:", device)
print("\nTraining Model A...")
results.append(run_model("Model A", blocks=2, embedding_dim=64))

print("\nTraining Model B...")
results.append(run_model("Model B", blocks=4, embedding_dim=128))

print("\n========== FINAL COMPARISON ==========")
print(
    f"{'Model':<10}{'Blocks':<9}{'Embed':<9}"
    f"{'Parameters':<15}{'Val Loss':<12}{'Time(s)':<12}"
)

for r in results:
    print(
        f"{r['model']:<10}{r['blocks']:<9}{r['embedding_dim']:<9}"
        f"{r['parameters']:<15,}{r['val_loss']:<12.4f}"
        f"{r['seconds']:<12.2f}"
    )
    if r["gpu_memory_mb"] is not None:
        print(f"  Peak GPU memory: {r['gpu_memory_mb']:.2f} MB")
```
