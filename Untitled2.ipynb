# Q5. Add a research memory

```python
import json
from datetime import datetime
from urllib.parse import urlparse, urlunparse, parse_qsl, urlencode
from typing import List
from pydantic import BaseModel

class Source(BaseModel):
    title: str
    url: str
    snippet: str
    query: str
    found_at: str

def normalize_url(url: str) -> str:
    """Make URL variants compare equal (case, trailing slash, #fragment, utm params)."""
    p = urlparse(url.strip())
    query = urlencode([(k, v) for k, v in parse_qsl(p.query) if not k.startswith("utm_")])
    path = p.path.rstrip("/")
    return urlunparse((p.scheme.lower(), p.netloc.lower().removeprefix("www."), path, "", query, ""))

class ResearchMemory:
    def __init__(self):
        self.sources: dict[str, Source] = {}  # key = normalized URL

    def add(self, title, url, snippet, query) -> bool:
        key = normalize_url(url)
        if not key or key in self.sources:
            return False  # duplicate
        self.sources[key] = Source(
            title=title, url=url, snippet=snippet, query=query,
            found_at=datetime.now().isoformat(timespec="seconds"),
        )
        return True

    def all(self) -> List[Source]:
        return list(self.sources.values())

    def save(self, path="research_memory.json"):
        with open(path, "w") as f:
            json.dump([s.model_dump() for s in self.all()], f, indent=2)

    def load(self, path="research_memory.json"):
        with open(path) as f:
            for item in json.load(f):
                self.sources[normalize_url(item["url"])] = Source(**item)

memory = ResearchMemory()
```

```python
def search_and_store(query: str) -> dict:
    results = search_web(query)
    new, dupes = [], 0
    for r in results:
        if memory.add(r["title"], r["url"], r["snippet"], query):
            new.append(r)
        else:
            dupes += 1
    return {"new_results": new, "duplicates_skipped": dupes}

TOOLS["search_web"] = search_and_store  # the agent loop now uses this version
```

```python
import requests
import trafilatura
from bs4 import BeautifulSoup

HEADERS = {"User-Agent": "Mozilla/5.0 (compatible; ResearchBot/1.0)"}
MAX_CHARS = 8000  # keep text small enough for the LLM

def fetch_page(url: str, timeout: int = 10) -> dict:
    """Fetch a page and extract its main text. Never raises; always returns a dict."""
    result = {"url": url, "ok": False, "text": "", "error": None}
    try:
        r = requests.get(url, headers=HEADERS, timeout=timeout)
        r.raise_for_status()

        ctype = r.headers.get("Content-Type", "")
        if "html" not in ctype.lower():
            result["error"] = f"Not an HTML page ({ctype or 'unknown type'})"
            return result

        html = r.text  # requests guesses encoding; bad bytes get replaced

        # 1st try: trafilatura (best at finding main article text)
        text = trafilatura.extract(html, include_comments=False, include_tables=False)

        # fallback: BeautifulSoup, strip junk tags and take all text
        if not text:
            soup = BeautifulSoup(html, "html.parser")
            for tag in soup(["script", "style", "nav", "footer", "header", "aside", "form"]):
                tag.decompose()
            text = " ".join(soup.get_text(separator=" ").split())

        if not text or len(text) < 100:
            result["error"] = "No readable text found"
            return result

        result["text"] = text[:MAX_CHARS]
        result["ok"] = True

    except requests.exceptions.Timeout:
        result["error"] = f"Timed out after {timeout}s"
    except requests.exceptions.HTTPError as e:
        result["error"] = f"HTTP error {e.response.status_code}"
    except requests.exceptions.SSLError:
        result["error"] = "SSL certificate error"
    except requests.exceptions.ConnectionError:
        result["error"] = "Could not connect"
    except requests.exceptions.InvalidURL:
        result["error"] = "Malformed URL"
    except requests.exceptions.RequestException as e:
        result["error"] = f"Request failed: {type(e).__name__}"
    except Exception as e:
        result["error"] = f"Parsing failed: {type(e).__name__}"

    return result
```

```python
memory.contents = {}  # url key -> extracted text

def fetch_and_store(url: str) -> dict:
    key = normalize_url(url)
    if key in memory.contents:
        return {"url": url, "ok": True, "text": memory.contents[key], "note": "from cache"}
    result = fetch_page(url)
    if result["ok"]:
        memory.contents[key] = result["text"]
    return result
```

# Q7. Extract evidence

```python
import json, time
from typing import List, Optional
from pydantic import BaseModel, Field, ValidationError
from google.genai import types

class Evidence(BaseModel):
    main_claims: List[str] = Field(description="Key claims made by the source")
    supporting_passages: List[str] = Field(description="Exact quotes from the text that support the claims")
    publication_date: Optional[str] = Field(default=None, description="Date if stated in the text, else null")
    uncertainties: List[str] = Field(description="Limitations, caveats, bias, or missing info")

class SourceEvidence(Evidence):
    source_url: str  # filled by our code, not the model

evidence_config = types.GenerateContentConfig(
    response_mime_type="application/json",
    response_schema=Evidence,
    temperature=0,
)

def extract_evidence(url: str, retries: int = 3) -> Optional[SourceEvidence]:
    key = normalize_url(url)
    text = memory.contents.get(key)
    if not text:
        print(f"No content stored for {url}")
        return None

    prompt = f"""Extract evidence from this web page text.
- main_claims: the key claims
- supporting_passages: EXACT quotes copied from the text (no paraphrasing)
- publication_date: only if clearly stated in the text, otherwise null
- uncertainties: limitations, bias, vague claims, or missing information
Use only the text below. Do not add outside knowledge.

TEXT:
{text}"""

    for attempt in range(retries):
        for model in MODELS:
            try:
                r = client.models.generate_content(model=model, contents=prompt, config=evidence_config)
                ev = Evidence.model_validate_json(r.text)
                # drop passages the model made up (not actually in the text)
                ev.supporting_passages = [p for p in ev.supporting_passages if p.strip() in text]
                return SourceEvidence(**ev.model_dump(), source_url=url)
            except ValidationError as e:
                print(f"Invalid output from {model}: {str(e)[:100]}")
            except Exception as e:
                print(f"{model} failed: {type(e).__name__}")
        time.sleep(5 * (attempt + 1))
    return None
```

```python
memory.evidence = {}

for src in memory.all():
    if normalize_url(src.url) not in memory.contents:
        fetch_and_store(src.url)          # fetch if not already fetched
    ev = extract_evidence(src.url)
    if ev:
        memory.evidence[normalize_url(src.url)] = ev

for ev in memory.evidence.values():
    print(json.dumps(ev.model_dump(), indent=2), "\n")

with open("evidence.json", "w") as f:
    json.dump([e.model_dump() for e in memory.evidence.values()], f, indent=2)
```

# Q8. Build a citation system

```python
import json, time
from typing import List
from pydantic import BaseModel, Field, ValidationError
from google.genai import types

# 1. Give every source with evidence a fixed ID
source_table = {}  # id -> SourceEvidence
for i, ev in enumerate(memory.evidence.values(), 1):
    source_table[i] = ev

def build_evidence_block():
    lines = []
    for sid, ev in source_table.items():
        lines.append(f"[S{sid}] URL: {ev.source_url}")
        lines.append(f"  Date: {ev.publication_date or 'unknown'}")
        lines.append("  Claims: " + " | ".join(ev.main_claims))
        lines.append("  Passages: " + " | ".join(ev.supporting_passages))
        lines.append("  Limits: " + " | ".join(ev.uncertainties))
    return "\n".join(lines)

# 2. Report format: every statement must carry source IDs and an exact quote
class Statement(BaseModel):
    text: str = Field(description="One factual statement for the report")
    source_ids: List[int] = Field(min_length=1, description="IDs of sources supporting it")
    quote: str = Field(description="Exact passage copied from a cited source")

class Report(BaseModel):
    title: str
    statements: List[Statement] = Field(min_length=1)

report_config = types.GenerateContentConfig(
    response_mime_type="application/json",
    response_schema=Report,
    temperature=0,
)

# 3. Validator: reject anything that cannot be traced
def validate_report(report: Report) -> List[str]:
    errors = []
    for n, st in enumerate(report.statements, 1):
        for sid in st.source_ids:
            if sid not in source_table:
                errors.append(f"Statement {n}: source S{sid} does not exist")
        valid = [s for s in st.source_ids if s in source_table]
        if valid and not any(st.quote.strip() in source_table[s].supporting_passages for s in valid):
            errors.append(f"Statement {n}: quote not found in cited sources")
    return errors

# 4. Generate, validate, retry with the errors if invalid
def write_report(question: str, max_attempts: int = 3) -> Report:
    feedback = ""
    for attempt in range(max_attempts):
        prompt = f"""Write a research report answering: {question}

Rules:
- Use ONLY the evidence below. No outside knowledge.
- Every statement must list source_ids and copy ONE exact passage as quote.
- If the evidence does not support something, leave it out.

EVIDENCE:
{build_evidence_block()}
{feedback}"""
        raw = None
        for model in MODELS:
            try:
                raw = client.models.generate_content(model=model, contents=prompt, config=report_config).text
                break
            except Exception as e:
                print(f"{model} failed: {type(e).__name__}")
        if not raw:
            time.sleep(5 * (attempt + 1))
            continue
        try:
            report = Report.model_validate_json(raw)
        except ValidationError as e:
            feedback = f"\nYour last output was invalid JSON/schema: {str(e)[:200]}"
            continue
        errs = validate_report(report)
        if not errs:
            return report
        print(f"Attempt {attempt+1} rejected: {errs}")
        feedback = "\nFix these problems from your last attempt:\n" + "\n".join(errs)
    raise RuntimeError("Could not produce a fully cited report.")

# 5. Render: citation numbers and URLs come from OUR table, not the model
def render_report(report: Report) -> str:
    out = [f"# {report.title}\n"]
    used = []
    for st in report.statements:
        marks = ""
        for sid in st.source_ids:
            if sid not in used:
                used.append(sid)
            marks += f"[{used.index(sid) + 1}]"
        out.append(f"- {st.text} {marks}\n  > \"{st.quote}\"")
    out.append("\n## Sources")
    for n, sid in enumerate(used, 1):
        ev = source_table[sid]
        out.append(f"[{n}] {ev.source_url} (date: {ev.publication_date or 'unknown'})")
    return "\n".join(out)

report = write_report("What are the latest uses of AI agents in robotics?")
print(render_report(report))
```

```python
import re

def norm(s):
    return re.sub(r"\s+", " ", s).strip().lower()

def statement_ok(st) -> bool:
    valid = [s for s in st.source_ids if s in source_table]
    if not valid:
        return False
    q = norm(st.quote)
    return any(q in norm(p) for s in valid for p in source_table[s].supporting_passages)

def write_report(question: str, max_attempts: int = 6) -> Report:
    for attempt in range(max_attempts):
        prompt = f"""Write a research report answering: {question}

Rules:
- Use ONLY the evidence below. No outside knowledge.
- Every statement must list source_ids and copy ONE exact passage as quote.
- If the evidence does not support something, leave it out.

EVIDENCE:
{build_evidence_block()}"""
        raw = None
        for model in MODELS:
            try:
                raw = client.models.generate_content(model=model, contents=prompt, config=report_config).text
                break
            except Exception as e:
                print(f"{model} failed: {type(e).__name__}")
        if not raw:
            wait = 10 * (attempt + 1)
            print(f"Servers busy, waiting {wait}s...")
            time.sleep(wait)
            continue
        try:
            report = Report.model_validate_json(raw)
        except ValidationError:
            print("Invalid schema, retrying...")
            continue

        good = [st for st in report.statements if statement_ok(st)]
        dropped = len(report.statements) - len(good)
        if good:
            if dropped:
                print(f"Dropped {dropped} statement(s) with untraceable quotes.")
            for st in good:
                st.source_ids = [s for s in st.source_ids if s in source_table]
            report.statements = good
            return report
        print("No valid statements, retrying...")
    raise RuntimeError("Could not produce a cited report. Try again in a few minutes.")
```

# Q.9 Add a verification step

```python
import json, time
from typing import List
from pydantic import BaseModel, Field, ValidationError
from google.genai import types

class UnsupportedClaim(BaseModel):
    statement_number: int
    reason: str = Field(description="Why the cited quote/evidence does not support the statement")

class Contradiction(BaseModel):
    description: str
    statement_numbers: List[int] = Field(default_factory=list)
    source_ids: List[int] = Field(description="Sources that disagree")

class Evaluation(BaseModel):
    unsupported_claims: List[UnsupportedClaim]
    contradictions: List[Contradiction]
    unanswered_questions: List[str]

eval_config = types.GenerateContentConfig(
    response_mime_type="application/json",
    response_schema=Evaluation,
    temperature=0,
)

# Questions the report should answer: use the plan's sub-questions if you have them
try:
    research_questions = plan.sub_questions
except NameError:
    research_questions = ["What are the latest uses of AI agents in robotics?"]

def build_report_block(report):
    lines = []
    for n, st in enumerate(report.statements, 1):
        lines.append(f"Statement {n}: {st.text}")
        lines.append(f"  Cited sources: {['S'+str(s) for s in st.source_ids]}")
        lines.append(f"  Quote: \"{st.quote}\"")
    return "\n".join(lines)

def evaluate_report(report, questions, retries=5) -> Evaluation:
    prompt = f"""You are a strict fact-checking reviewer. Do not rewrite the report.

Check the REPORT against the EVIDENCE and answer:
1. unsupported_claims: statements whose cited quote/evidence does NOT actually support what the statement says (overreach, wrong numbers, added details). Give the statement number and reason.
2. contradictions: places where sources disagree with each other or with a statement. Give a description, the statement numbers, and the source IDs involved.
3. unanswered_questions: research questions below that the report does not answer or only answers weakly.

Use only the evidence given. If nothing is found for a category, return an empty list.

RESEARCH QUESTIONS:
{json.dumps(questions, indent=2)}

REPORT:
{build_report_block(report)}

EVIDENCE:
{build_evidence_block()}"""

    for attempt in range(retries):
        for model in MODELS:
            try:
                r = client.models.generate_content(model=model, contents=prompt, config=eval_config)
                ev = Evaluation.model_validate_json(r.text)
                # code-level validation: drop references to things that don't exist
                n = len(report.statements)
                ev.unsupported_claims = [u for u in ev.unsupported_claims if 1 <= u.statement_number <= n]
                for c in ev.contradictions:
                    c.statement_numbers = [s for s in c.statement_numbers if 1 <= s <= n]
                    c.source_ids = [s for s in c.source_ids if s in source_table]
                return ev
            except ValidationError:
                print(f"{model}: invalid output, retrying...")
            except Exception as e:
                print(f"{model} failed: {type(e).__name__}")
        wait = 10 * (attempt + 1)
        print(f"Waiting {wait}s...")
        time.sleep(wait)
    raise RuntimeError("Evaluation failed, try again in a few minutes.")

def render_evaluation(ev: Evaluation) -> str:
    out = ["## Verification\n", "### Unsupported claims"]
    out += [f"- Statement {u.statement_number}: {u.reason}" for u in ev.unsupported_claims] or ["- None found"]
    out.append("\n### Contradictory evidence")
    out += [f"- {c.description} (statements {c.statement_numbers}, sources {['S'+str(s) for s in c.source_ids]})"
            for c in ev.contradictions] or ["- None found"]
    out.append("\n### Unanswered research questions")
    out += [f"- {q}" for q in ev.unanswered_questions] or ["- None found"]
    return "\n".join(out)

evaluation = evaluate_report(report, research_questions)
print(render_evaluation(evaluation))
```

```python
import os, re, json, time
from datetime import datetime
from typing import List, Optional
from urllib.parse import urlparse, urlunparse, parse_qsl, urlencode

import requests, trafilatura
from bs4 import BeautifulSoup
from ddgs import DDGS
from pydantic import BaseModel, ValidationError
from google import genai
from google.genai import types
from google.colab import userdata

os.environ["GEMINI_API_KEY"] = userdata.get("GEMINI_API_KEY")
client = genai.Client(http_options=types.HttpOptions(timeout=60000))
MODELS = ["gemini-flash-lite-latest", "gemini-flash-latest"]
MAX_SOURCES = 5
HEADERS = {"User-Agent": "Mozilla/5.0 (compatible; ResearchBot/1.0)"}

# ---------- Schemas ----------
class ResearchPlan(BaseModel):
    objective: str
    sub_questions: List[str]
    search_queries: List[str]
    success_criteria: List[str]

class Evidence(BaseModel):
    main_claims: List[str]
    supporting_passages: List[str]
    publication_date: Optional[str]
    uncertainties: List[str]

class SourceEvidence(Evidence):
    source_url: str

class Statement(BaseModel):
    text: str
    source_ids: List[int]
    quote: str

class Report(BaseModel):
    title: str
    statements: List[Statement]

class UnsupportedClaim(BaseModel):
    statement_number: int
    reason: str

class Contradiction(BaseModel):
    description: str
    statement_numbers: List[int]
    source_ids: List[int]

class Evaluation(BaseModel):
    unsupported_claims: List[UnsupportedClaim]
    contradictions: List[Contradiction]
    unanswered_questions: List[str]

# ---------- LLM call with retry, fallback, validation ----------
def llm_json(prompt, schema, retries=5):
    config = types.GenerateContentConfig(
        response_mime_type="application/json", response_schema=schema, temperature=0)
    for attempt in range(retries):
        for model in MODELS:
            try:
                r = client.models.generate_content(model=model, contents=prompt, config=config)
                return schema.model_validate_json(r.text)
            except ValidationError:
                print(f"  {model}: invalid output, retrying")
            except Exception as e:
                print(f"  {model} failed: {type(e).__name__}")
        time.sleep(8 * (attempt + 1))
    return None

# ---------- Helpers ----------
def norm(s):
    return re.sub(r"\s+", " ", s).strip().lower()

def normalize_url(url):
    try:
        p = urlparse(url.strip())
        q = urlencode([(k, v) for k, v in parse_qsl(p.query) if not k.startswith("utm_")])
        return urlunparse((p.scheme.lower(), p.netloc.lower().removeprefix("www."), p.path.rstrip("/"), "", q, ""))
    except Exception:
        return ""

# ---------- Search tool ----------
def search_web(query, max_results=5):
    try:
        res = DDGS().text(query, max_results=max_results)
    except Exception as e:
        print(f"  search failed: {type(e).__name__}")
        return []
    return [{"title": r.get("title", ""), "url": r.get("href", ""), "snippet": r.get("body", "")} for r in res]

# ---------- Page reader ----------
def fetch_page(url, timeout=10):
    out = {"ok": False, "text": "", "error": None}
    try:
        r = requests.get(url, headers=HEADERS, timeout=timeout)
        r.raise_for_status()
        if "html" not in r.headers.get("Content-Type", "").lower():
            out["error"] = "Not HTML"
            return out
        text = trafilatura.extract(r.text, include_comments=False, include_tables=False)
        if not text:
            soup = BeautifulSoup(r.text, "html.parser")
            for t in soup(["script", "style", "nav", "footer", "header", "aside", "form"]):
                t.decompose()
            text = " ".join(soup.get_text(separator=" ").split())
        if not text or len(text) < 100:
            out["error"] = "No readable text"
            return out
        out.update(ok=True, text=text[:8000])
    except requests.exceptions.Timeout:
        out["error"] = "Timeout"
    except requests.exceptions.HTTPError as e:
        out["error"] = f"HTTP {e.response.status_code}"
    except requests.exceptions.RequestException as e:
        out["error"] = type(e).__name__
    except Exception as e:
        out["error"] = f"Parse error: {type(e).__name__}"
    return out
```

```python
class ResearchAgent:
    def __init__(self, question):
        self.question = question
        self.plan = None
        self.sources, self.texts, self.failed = {}, {}, []
        self.evidence, self.ids = [], {}
        self.report, self.evaluation = None, None

    # 1. PLAN
    def make_plan(self):
        self.plan = llm_json(
            f"Create a research plan for: {self.question}\n"
            "Return objective, sub_questions (3-5), search_queries (3-4), success_criteria.", ResearchPlan)
        if not self.plan or not self.plan.search_queries:
            raise RuntimeError("Planning failed")

    # 2. SEARCH (deduplicated)
    def search(self):
        for q in self.plan.search_queries[:4]:
            for r in search_web(q, 5):
                key = normalize_url(r["url"])
                if key and key not in self.sources:
                    self.sources[key] = {**r, "query": q}
            time.sleep(1)
        if not self.sources:
            raise RuntimeError("No search results")

    # 3. READ
    def read(self):
        for key, s in self.sources.items():
            if len(self.texts) >= MAX_SOURCES:
                break
            res = fetch_page(s["url"])
            if res["ok"]:
                self.texts[key] = res["text"]
            else:
                self.failed.append((s["url"], res["error"]))
        if not self.texts:
            raise RuntimeError("No pages could be read")

    # 4. EXTRACT EVIDENCE
    def extract(self):
        for key, text in self.texts.items():
            url = self.sources[key]["url"]
            ev = llm_json(f"""Extract evidence from this web page text.
- main_claims: key claims
- supporting_passages: EXACT quotes copied from the text
- publication_date: only if clearly stated, otherwise null
- uncertainties: limitations, bias, vague claims, missing info
Use only the text below.

TEXT:
{text}""", Evidence)
            time.sleep(2)
            if not ev:
                continue
            ev.supporting_passages = [p for p in ev.supporting_passages if norm(p) and norm(p) in norm(text)]
            if ev.supporting_passages:
                self.evidence.append(SourceEvidence(**ev.model_dump(), source_url=url))
        self.ids = {i: e for i, e in enumerate(self.evidence, 1)}
        if not self.ids:
            raise RuntimeError("No usable evidence extracted")

    def evidence_block(self):
        lines = []
        for sid, e in self.ids.items():
            lines += [f"[S{sid}] URL: {e.source_url}",
                      f"  Date: {e.publication_date or 'unknown'}",
                      "  Claims: " + " | ".join(e.main_claims),
                      "  Passages: " + " | ".join(e.supporting_passages),
                      "  Limits: " + " | ".join(e.uncertainties)]
        return "\n".join(lines)

    # 5. REPORT (only traceable statements survive)
    def write_report(self, attempts=3):
        for _ in range(attempts):
            rep = llm_json(f"""Write a research report answering: {self.question}
Rules:
- Use ONLY the evidence below. No outside knowledge.
- Every statement needs source_ids and ONE exact passage copied as quote.
- Leave out anything the evidence does not support.

EVIDENCE:
{self.evidence_block()}""", Report)
            if not rep:
                continue
            good = []
            for st in rep.statements:
                ids = [s for s in st.source_ids if s in self.ids]
                q = norm(st.quote)
                if ids and q and any(q in norm(p) for s in ids for p in self.ids[s].supporting_passages):
                    st.source_ids = ids
                    good.append(st)
            if good:
                rep.statements = good
                self.report = rep
                return
        raise RuntimeError("No traceable report produced")

    # 6. VERIFY (separate critic stage)
    def verify(self):
        n = len(self.report.statements)
        block = "\n".join(f'Statement {i}: {s.text}\n  Sources: {["S"+str(x) for x in s.source_ids]}\n  Quote: "{s.quote}"'
                          for i, s in enumerate(self.report.statements, 1))
        ev = llm_json(f"""You are a strict fact-checking reviewer. Do not rewrite the report.
1. unsupported_claims: statements the cited quote does NOT actually support (overreach, wrong details).
2. contradictions: where sources disagree with each other or a statement.
3. unanswered_questions: research questions the report does not answer.
Use only the evidence. Return empty lists if nothing is found.

RESEARCH QUESTIONS:
{json.dumps(self.plan.sub_questions)}

REPORT:
{block}

EVIDENCE:
{self.evidence_block()}""", Evaluation)
        if not ev:
            print("  verification unavailable")
            return
        ev.unsupported_claims = [u for u in ev.unsupported_claims if 1 <= u.statement_number <= n]
        for c in ev.contradictions:
            c.statement_numbers = [s for s in c.statement_numbers if 1 <= s <= n]
            c.source_ids = [s for s in c.source_ids if s in self.ids]
        self.evaluation = ev

    # RENDER
    def render(self):
        flagged = {u.statement_number: u.reason for u in self.evaluation.unsupported_claims} if self.evaluation else {}
        used = []
        L = [f"# {self.report.title}", f"*Question: {self.question}*",
             f"*Generated: {datetime.now():%Y-%m-%d %H:%M}*", "",
             f"**Objective:** {self.plan.objective}", "", "## Findings"]
        for n, st in enumerate(self.report.statements, 1):
            marks = ""
            for sid in st.source_ids:
                if sid not in used:
                    used.append(sid)
                marks += f"[{used.index(sid) + 1}]"
            flag = f" ⚠️ *Flagged by verifier: {flagged[n]}*" if n in flagged else ""
            L += [f"{n}. {st.text} {marks}{flag}", f'   > "{st.quote}"']
        L += ["", "## Sources"]
        for i, sid in enumerate(used, 1):
            e = self.ids[sid]
            L.append(f"[{i}] {e.source_url} (date: {e.publication_date or 'unknown'})")
        L += ["", "## Verification"]
        if self.evaluation:
            ev = self.evaluation
            L.append("**Unsupported claims**")
            L += [f"- Statement {u.statement_number}: {u.reason}" for u in ev.unsupported_claims] or ["- None found"]
            L.append("\n**Contradictory evidence**")
            L += [f"- {c.description} (statements {c.statement_numbers}, sources {['S'+str(s) for s in c.source_ids]})"
                  for c in ev.contradictions] or ["- None found"]
            L.append("\n**Unanswered research questions**")
            L += [f"- {q}" for q in ev.unanswered_questions] or ["- None found"]
        else:
            L.append("Verification was unavailable (API error).")
        L += ["", "**Success criteria from plan (check manually):**"] + [f"- {c}" for c in self.plan.success_criteria]
        L += ["", "## Run info",
              f"- Sources found: {len(self.sources)}",
              f"- Pages read: {len(self.texts)}",
              f"- Pages failed: {len(self.failed)} " + (str([f"{e}" for _, e in self.failed]) if self.failed else ""),
              f"- Sources with usable evidence: {len(self.ids)}"]
        return "\n".join(L)

    def run(self):
        for name, fn in [("Plan", self.make_plan), ("Search", self.search), ("Read", self.read),
                         ("Extract evidence", self.extract), ("Report", self.write_report), ("Verify", self.verify)]:
            t = time.time()
            print(f"[{name}] ...")
            fn()
            print(f"[{name}] done in {time.time() - t:.0f}s")
        return self.render()
```

```python
QUESTIONS = [
    "What are the main approaches to reducing hallucinations in large language models?",
    "How does retrieval-augmented generation (RAG) work and what are its limitations?",
]

OUT = "research_output"
os.makedirs(f"{OUT}/reports", exist_ok=True)

for i, q in enumerate(QUESTIONS, 1):
    print(f"\n{'='*60}\nQuestion {i}: {q}\n{'='*60}")
    try:
        agent = ResearchAgent(q)
        md = agent.run()
        with open(f"{OUT}/reports/report_{i}.md", "w") as f:
            f.write(md)
        with open(f"{OUT}/reports/report_{i}.json", "w") as f:
            json.dump({
                "question": q,
                "plan": agent.plan.model_dump(),
                "evidence": [e.model_dump() for e in agent.evidence],
                "report": agent.report.model_dump(),
                "evaluation": agent.evaluation.model_dump() if agent.evaluation else None,
            }, f, indent=2)
        print("\n" + md)
    except Exception as e:
        print(f"Question {i} failed: {e}")
        with open(f"{OUT}/reports/report_{i}_FAILED.txt", "w") as f:
            f.write(f"{q}\nFailed: {e}\n")
```

```python
README = """# Research Agent

A pipeline that plans, searches, reads, extracts evidence, writes a cited report, and verifies it.

## Pipeline
Plan -> Search -> Read -> Extract evidence -> Write cited report -> Verify -> Save

## Tools used
1. Python
2. LLM API (Google Gemini via google-genai)
3. Pydantic (validates every model output)
4. Search API (DuckDuckGo via the ddgs package)
5. requests (fetching pages)
6. BeautifulSoup (HTML fallback parsing; trafilatura is the main extractor)
7. Git and GitHub (version control and sharing)

## How citations are protected
- The model can only cite source IDs (S1, S2...) from a table built by the code.
- Each statement must include a quote that exactly matches a passage extracted from the cited page.
- Statements with unknown IDs or non-matching quotes are dropped.
- URLs in the final report come from stored data, never from model text.
- A separate verifier stage flags unsupported claims, contradictions, and unanswered questions.

## Limitations
- Free search (DuckDuckGo) is unofficial, can be rate-limited or return empty results, and gives no control over ranking or freshness.
- Only static HTML pages are read. Pages that need JavaScript, logins, or payment, and PDFs, fail.
- Page text is cut at 8000 characters, so content later in long pages is missed.
- Only up to 5 pages are read per question, so this is a quick survey, not a systematic review.
- Exact-quote matching proves a quote exists in the page. It does not prove the claim is true or that the statement fully matches the quote.
- The verifier is also an LLM and can miss problems or flag good statements. Flagged items are marked, not removed.
- Source quality is not scored. Blogs, vendor marketing, and low-quality sites are treated like any other source.
- Publication dates are only captured when stated in the page text.
- Contradiction detection is limited to the few sources read.
- Free API tiers get overloaded or rate-limited (503/429 errors), so runs can fail or be slow.
- Reports should be read by a human before being trusted.

## Setup
Store the API key as a Colab secret named GEMINI_API_KEY. Never put keys in code or commit them to GitHub.
"""
with open(f"{OUT}/README.md", "w") as f:
    f.write(README)

import shutil
from google.colab import files
shutil.make_archive("research_agent_output", "zip", OUT)
files.download("research_agent_output.zip")
```
